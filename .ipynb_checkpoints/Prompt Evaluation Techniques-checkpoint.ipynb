Writing an effective prompt is only the beginning of Prompt Engineering. A prompt may generate an excellent response for one input but produce inaccurate, incomplete, or inconsistent results for another.

How do we determine whether a prompt is actually performing well?

This is where **Prompt Evaluation Techniques** become important. Prompt Evaluation involves systematically testing prompts against predefined inputs and measuring their outputs using evaluation metrics. Instead of relying entirely on manual observation, we can use Python to compare prompt variations, measure response quality, and identify which prompts perform more reliably.

## 📦 Import Required Libraries

```python
import pandas as pd
from transformers import pipeline
from sentence_transformers import SentenceTransformer, util
from sklearn.metrics import accuracy_score
```

These libraries will help us generate responses, compare predictions, and evaluate their quality.

---

## 🤖 Step 1 — Load the Language Models

First, we'll load an instruction-following language model for generating responses. We'll also load a sentence embedding model for measuring semantic similarity.

```python
generator = pipeline("text2text-generation", model="google/flan-t5-base")
embedding_model = SentenceTransformer("sentence-transformers/all-MiniLM-L6-v2")
```

**Why use two models?**

The FLAN-T5 model generates responses to our prompts. The Sentence Transformer model converts text into numerical embeddings, allowing us to compare the semantic meaning of generated responses with expected answers. This helps us evaluate both correctness and meaning.

---

## 📝 Step 2 — Create an Evaluation Dataset

A prompt should be tested against multiple examples rather than a single input. We'll create a small sentiment classification dataset.

```python
evaluation_data = [
    {
        "text": "The application is incredibly fast and easy to use.",
        "expected": "Positive"
    },
    {
        "text": "The latest update keeps crashing my application.",
        "expected": "Negative"
    },
    {
        "text": "The software was installed yesterday.",
        "expected": "Neutral"
    },
    {
        "text": "Customer support resolved my issue immediately.",
        "expected": "Positive"
    },
    {
        "text": "The interface is confusing and difficult to navigate.",
        "expected": "Negative"
    },
    {
        "text": "The meeting is scheduled for Monday.",
        "expected": "Neutral"
    }
]

df = pd.DataFrame(evaluation_data)
print(df)
```

Each record contains an input and its expected output. These expected outputs serve as **reference answers** for evaluating prompt performance.

In production, evaluation datasets should contain more diverse examples, including ambiguous inputs and challenging edge cases.

---

## 🔍 Step 3 — Define Different Prompt Variations

Now we'll create two prompts that perform the same task but use different instructions.

**Prompt A — Basic Instruction**

```python
def basic_prompt(text):
    return f"Classify the sentiment of this text: {text}"
```

**Prompt B — Structured Instruction**

```python
def structured_prompt(text):
    return f"""
Classify the sentiment of the following text.

Choose exactly one category:
Positive, Negative, Neutral

Return only the category name.

Text: {text}
"""
```

Both prompts ask the model to classify sentiment. However, Prompt B explicitly defines the available categories and expected output format. Our goal is to determine whether these additional instructions improve performance.

---

## ⚙️ Step 4 — Generate Model Responses

We'll create a reusable function to execute prompts.

```python
def generate_response(prompt):
    response = generator(
        prompt,
        max_new_tokens=20,
        do_sample=False
    )
    return response[0]["generated_text"].strip()
```

Setting `do_sample=False` uses deterministic decoding, making comparisons more repeatable under the same model and execution conditions. Now we can evaluate both prompt variations.

```python
df["Basic_Prediction"] = df["text"].apply(lambda text: generate_response(basic_prompt(text)))
df["Structured_Prediction"] = df["text"].apply(lambda text: generate_response(structured_prompt(text)))
```

Our dataset now contains predictions generated by both prompts.

---

## 📊 Step 5 — Evaluate Classification Accuracy

Accuracy measures how frequently the generated predictions match the expected labels.

The formula is:

**Accuracy = Correct Predictions / Total Predictions**

We'll normalise the generated labels before comparing them.

```python
def normalize_label(text):
    return text.strip().lower()

expected = df["expected"].apply(normalize_label)
basic_predictions = df["Basic_Prediction"].apply(normalize_label)
structured_predictions = df["Structured_Prediction"].apply(normalize_label)
```

Calculate accuracy:

```python
basic_accuracy = accuracy_score(expected, basic_predictions)
structured_accuracy = accuracy_score(expected, structured_predictions)

print(f"Basic Prompt Accuracy: {basic_accuracy:.2%}")
print(f"Structured Prompt Accuracy: {structured_accuracy:.2%}")
```

This gives us a measurable way to compare the two prompts. For example, if the structured prompt produces more correct classifications, its accuracy score will be higher. However, this is only one evaluation metric.

---

## 🧠 Step 6 — Evaluate Semantic Similarity

Exact matching works well for classification tasks, but many LLM applications generate free-form text.

For example:

**Expected Answer:**

***Machine Learning allows computers to learn patterns from data.***

**Generated Answer:**

***Machine Learning enables systems to identify patterns using data.***

These sentences use different words but communicate similar meanings. Exact string matching would treat them as different. Semantic similarity provides another perspective.

```python
expected_answer = "Machine Learning allows computers to learn patterns from data."
generated_answer = "Machine Learning enables systems to identify patterns using data."
```

Generate embeddings:

```python
expected_embedding = embedding_model.encode(expected_answer, convert_to_tensor=True)
generated_embedding = embedding_model.encode(generated_answer, convert_to_tensor=True)
```

Calculate cosine similarity:

```python
similarity = util.cos_sim(expected_embedding, generated_embedding).item()
print(f"Semantic Similarity: {similarity:.4f}")
```

A higher similarity score generally indicates that the responses are closer in semantic meaning. **Important:** Semantic similarity does not guarantee factual correctness. Two sentences can be semantically similar while containing different facts. It should therefore be treated as a supporting metric rather than a complete measure of response quality.

---

## 🔄 Step 7 — Compare Prompt Performance

Let's summarise our evaluation results.

```python
results = pd.DataFrame({
    "Prompt": ["Basic Prompt", "Structured Prompt"],
    "Accuracy": [basic_accuracy, structured_accuracy]
})
print(results)
```

We can also identify which prompt performed better.

```python
best_prompt = results.loc[
    results["Accuracy"].idxmax(),
    "Prompt"
]
print(f"Best Performing Prompt: {best_prompt}")
```

If both prompts achieve the same accuracy, this selects the first one. In a larger evaluation framework, ties should be reported explicitly and resolved using additional criteria.

---

## 📈 Step 8 — Measure Output Consistency

A good classification prompt should produce outputs that follow the required format.

For our example, the valid responses are:

```python
valid_labels = {
    "positive",
    "negative",
    "neutral"
}
```

We can measure how frequently each prompt produces a valid category.

```python
def calculate_consistency(predictions):
    valid_count = sum(
        prediction in valid_labels
        for prediction in predictions
    )
    return valid_count / len(predictions)
```

Evaluate both prompts:

```python
basic_consistency = calculate_consistency(basic_predictions)
structured_consistency = calculate_consistency(structured_predictions)

print(f"Basic Prompt Consistency: {basic_consistency:.2%}")
print(f"Structured Prompt Consistency: {structured_consistency:.2%}")
```

Here, consistency means **output-format compliance**, not stability across repeated generations. A model might return the wrong sentiment while still following the required output format. This demonstrates why accuracy and consistency should be measured separately.

---

## 🧪 Step 9 — Build a Reusable Evaluation Function

We can combine our evaluation logic into a reusable function.

```python
def evaluate_prompt(prompt_function, dataset):
    predictions = []

    for text in dataset["text"]:
        prompt = prompt_function(text)
        prediction = generate_response(prompt)
        predictions.append(normalize_label(prediction))

    expected = dataset["expected"].apply(normalize_label)

    accuracy = accuracy_score(
        expected,
        predictions
    )

    consistency = calculate_consistency(predictions)

    return {
        "accuracy": accuracy,
        "consistency": consistency,
        "predictions": predictions
    }
```

Now we can evaluate any compatible prompt template.

```python
basic_results = evaluate_prompt(basic_prompt, df)
structured_results = evaluate_prompt(structured_prompt, df)

print("Basic Prompt:", basic_results["accuracy"])
print("Structured Prompt:", structured_results["accuracy"])
```

This reusable approach makes it easier to extend our experiments with additional prompts and evaluation datasets.

## 📊 Understanding Prompt Evaluation Metrics

Different LLM applications require different evaluation techniques.

| Evaluation Technique | What It Measures | Common Use Cases |
| --- | --- | --- |
| Exact Match | Whether output matches the expected answer | Classification, structured extraction |
| Accuracy | Percentage of correct predictions | Sentiment analysis, categorisation |
| Semantic Similarity | Similarity in meaning between responses | Summarisation, question answering |
| Format Compliance | Whether output follows required formatting | JSON generation, classification |
| Human Evaluation | Quality judged by human reviewers | Creative writing, complex responses |
| LLM-as-a-Judge | Quality assessed by another LLM | Open-ended response evaluation |
| Latency | Time required to generate a response | Production applications |
| Token Usage | Tokens consumed during generation | Cost and efficiency monitoring |

No single metric can fully evaluate every LLM application. For example, semantic similarity may be useful for summarisation, but accuracy and exact matching are more appropriate for sentiment classification. The evaluation technique should match the task being performed.

## 🧠 Key Takeaways

- Prompt Evaluation measures how effectively prompts perform across predefined test cases.
- Evaluation datasets help compare prompts using consistent inputs and expected outputs.
- Accuracy measures classification correctness.
- Semantic similarity measures closeness in meaning but does not guarantee factual accuracy.
- Format compliance helps determine whether responses follow required output structures.
- Reusable evaluation functions simplify comparisons between prompt variations.
- Production evaluation should include diverse datasets, relevant quality metrics, and repeated testing where appropriate.
