**PYTHON SCRIPT TO *UNDERSTAND HOW AI CONVERTS TEXT INTO NUMERICAL VECTORS FOR SEMANTIC SEARCH*. 🐍🔍**

Large Language Models work with text, but similarity-search systems need numerical representations to compare information efficiently.

**Vector Embeddings** solve this problem by representing text as numerical vectors that capture aspects of its semantic meaning.

In this project, we'll explore how embeddings work, generate them using Python, and measure semantic similarity between sentences.

---

## 📦 Step 1 — Install Required Libraries

We'll use Sentence Transformers to generate embeddings and NumPy for numerical operations.

```
pip install sentence-transformers numpy
```

````python
import numpy as np
from sentence_transformers import SentenceTransformer
````

---

## 🧠 Step 2 — What Are Vector Embeddings?

A vector embedding is a numerical representation of text generated by a trained model.

For example:

```
"Learn Python" → [0.21, -0.45, 0.78, ...]
```

Each number represents a learned feature within a multidimensional vector space.

The important idea is that **texts with similar meanings generally produce vectors that are closer together**.

Consider these sentences:

- "How can I learn Python?"
- "What's the best way to study Python?"
- "The weather is sunny today."

The first two sentences express similar meanings, while the third discusses something unrelated.

An embedding model should generally assign a higher similarity score to the first two sentences.

---

## 🤖 Step 3 — Generate Text Embeddings

We'll use the pre-trained `all-MiniLM-L6-v2` model, which produces **384-dimensional embeddings**.

````python
model = SentenceTransformer("sentence-transformers/all-MiniLM-L6-v2")

sentences = [
    "How can I learn Python?",
    "What's the best way to study Python?",
    "The weather is sunny today."
]

embeddings = model.encode(sentences, convert_to_numpy=True)

print("Embedding Shape:", embeddings.shape)
print("First Embedding:", embeddings[0][:5])
````

**Expected shape:**

```
Embedding Shape: (3, 384)
```

Each sentence is represented by 384 numerical values.

We display only the first five values of the first embedding to keep the output readable.

---

## 📐 Step 4 — Measure Semantic Similarity

Once we have embeddings, we can compare them using **cosine similarity**.

Cosine similarity measures the angle between two vectors:

$$
\text{Cosine Similarity}(A,B) = \frac{A \cdot B}{\|A\|\|B\|}
$$

Its mathematical range is **-1 to 1**:

- **1:** Vectors point in the same direction.
- **0:** Vectors are perpendicular.
- **1:** Vectors point in opposite directions.

For text embeddings, higher scores generally indicate greater semantic similarity. However, a high score does not guarantee that two statements are factually equivalent.

Let's calculate similarity using Python.

````python
normalized = embeddings / np.linalg.norm(embeddings, axis=1, keepdims=True)
similarity_matrix = normalized @ normalized.T

print("Similarity Matrix:")
print(np.round(similarity_matrix, 3))
````

The similarity matrix compares every sentence against every other sentence.

We generally expect the first two sentences to have a higher similarity score than either has with the weather-related sentence.

---

## 🔍 Step 5 — Find the Most Similar Sentence

Let's identify which sentence is most similar to our first sentence.

````python
scores = similarity_matrix[0].copy()
scores[0] = -np.inf

best_match = np.argmax(scores)

print("Original:", sentences[0])
print("Best Match:", sentences[best_match])
print(f"Similarity Score: {scores[best_match]:.4f}")
````

We exclude the original sentence from the comparison because its similarity with itself is always approximately 1.

This demonstrates the basic principle behind **semantic search**: finding relevant information based on meaning rather than exact keyword matches.

## 🔗 Step 6 — How Embeddings Work in RAG

In a Retrieval-Augmented Generation system, embeddings connect user questions with relevant documents.

The process is straightforward:

```
Documents → Embeddings → Vector Store
                              ↑
User Query → Query Embedding → Similarity Search
                              ↓
                       Relevant Documents
                              ↓
                         LLM Response
```

The workflow involves:

1. Converting document chunks into embeddings.
2. Storing those embeddings in a vector store such as FAISS.
3. Converting a user's question into an embedding using the same model.
4. Finding the most similar document embeddings.
5. Passing the retrieved text to an LLM as context for generating an answer.

**Important:** Embeddings don't generate answers. They help the retrieval system identify relevant information.

## 🌍 Real-World Applications

Vector embeddings are useful in several AI applications:

- **Semantic Search:** Finding documents based on meaning.
- **RAG Systems:** Retrieving relevant context for LLM responses.
