**PYTHON SCRIPT TO *BUILD AN INTERACTIVE PLAYGROUND TO EXPERIMENT WITH PROMPTS AND LLM SETTINGS. 🧠🎛️🤖***

Prompt Engineering involves experimentation. Small changes to instructions, temperature, response length, or context can significantly affect the output generated by a Large Language Model.

Instead of modifying Python code every time we want to test a prompt, we can build an interactive **Prompt Playground** where prompts and generation settings can be changed directly from a simple user interface.

In this project, we'll build a Prompt Playground using **Python, Hugging Face Transformers, and Streamlit**.

## 📦 Import Required Libraries

````python
import streamlit as st
from transformers import pipeline
````

Streamlit will provide the interactive interface, while Hugging Face Transformers will handle text generation.

---

## 🤖 Load the Language Model

Loading an LLM can take time, so we'll cache the model instead of loading it again after every Streamlit interaction.

````python
@st.cache_resource
def load_model():
    return pipeline(
        "text-generation",
        model="google/gemma-2b-it"
    )

generator = load_model()
````

`st.cache_resource` keeps the model available across Streamlit reruns, making the playground much more efficient.

---

## 🖥️ Create the Playground Interface

Start by adding a title and short description.

````python
st.title("🧠 Prompt Playground")

st.write(
    "Experiment with prompts and generation settings "
    "using an instruction-following language model."
)
````

This gives us the foundation for our interactive application.

---

## ✍️ Add a Prompt Editor

Users need an area where they can freely write and modify prompts.

````python
prompt = st.text_area(
    "Enter your prompt",
    value="Explain Machine Learning to a beginner.",
    height=180
)
````

Unlike a fixed Python string, the prompt can now be changed directly from the browser.

This makes it easy to experiment with different instructions without editing the application code.

---

## 🎛️ Control the Generation Settings

Prompt wording isn't the only thing that influences an LLM response. Generation parameters also affect how the model behaves.

We'll add controls for response length and creativity.

````python
max_tokens = st.slider(
    "Maximum New Tokens",
    min_value=20,
    max_value=300,
    value=120
)

temperature = st.slider(
    "Temperature",
    min_value=0.1,
    max_value=1.5,
    value=0.7,
    step=0.1
)
````

`max_new_tokens` controls the maximum length of the generated response.

`temperature` controls randomness. Lower values generally produce more predictable responses, while higher values introduce more variation.

---

## 🚀 Generate a Response

Now we'll add a button that sends the current prompt and settings to the model.

````python
if st.button("Generate Response"):
    response = generator(
        prompt,
        max_new_tokens=max_tokens,
        temperature=temperature,
        do_sample=True
    )

    output = response[0]["generated_text"]

    st.subheader("Generated Response")
    st.write(output)
````

Every time the button is clicked, the playground uses the latest prompt and generation settings.

---

## 🎯 Add Prompt Presets

A Prompt Playground becomes more useful when users can quickly experiment with different types of prompts.

````python
preset = st.selectbox(
    "Prompt Preset",
    [
        "Custom",
        "Explain a Concept",
        "Summarize Text",
        "Generate Ideas",
        "Classify Text"
    ]
)
````

Presets can provide starting points for common Prompt Engineering tasks while still allowing the user to modify the generated prompt.
