Large Language Models can often solve complex problems with a single prompt. However, for tasks involving logic, mathematics, planning, or multi-step reasoning, simply asking for the final answer may not produce the best results.

**Chain-of-Thought (CoT) Prompting** is a Prompt Engineering technique that encourages the model to reason through a problem step by step before arriving at its final answer. By explicitly asking the model to explain its thinking process, we can often improve the quality and accuracy of responses for complex tasks.

In this project, we'll compare standard prompting with Chain-of-Thought prompting using a Hugging Face instruction-following language model.

We'll learn how to:

- Understand Chain-of-Thought Prompting
- Compare normal prompts with reasoning prompts
- Encourage step-by-step reasoning
- Build reusable reasoning prompts
- Explore practical reasoning tasks

## 📦 Import Required Libraries

````python
from transformers import pipeline
````

We'll use an instruction-following language model to observe how responses change when reasoning is explicitly requested.

---

## 🤖 Load an Instruction-Following Model

````python
generator = pipeline(
    "text-generation",
    model="google/gemma-2b-it"
)
````

Instruction-tuned models are capable of following reasoning-based prompts and producing detailed step-by-step explanations.

---

## ❓ Ask a Question Normally

````python
normal_prompt = """
A store sells notebooks for $5 each.

If you buy 8 notebooks, how much do you pay?
"""
````

This prompt asks only for the answer, without encouraging the model to explain its reasoning.

---

## 🚀 Generate the Normal Response

````python
response = generator(
    normal_prompt,
    max_new_tokens=80,
    do_sample=False
)

print(response[0]["generated_text"])
````

The model typically provides the answer directly with little or no explanation.

---

## 🧠 Create a Chain-of-Thought Prompt

````python
cot_prompt = """
A store sells notebooks for $5 each.

If you buy 8 notebooks, how much do you pay?

Let's think step by step.
"""
````

The phrase **"Let's think step by step."** encourages the model to reason through the problem before producing its final answer.

---

## 🚀 Generate the Chain-of-Thought Response

````python
response = generator(
    cot_prompt,
    max_new_tokens=150,
    do_sample=False
)

print(response[0]["generated_text"])
````

Notice how the model explains each intermediate step before reaching the final result.

---

## 🔄 Compare Multiple Reasoning Tasks

````python
questions = [
    "If a train travels 60 km in 1 hour, how far will it travel in 5 hours?",
    "John has 12 apples. He gives 5 away and buys 3 more. How many apples does he have now?",
    "A rectangle has a length of 8 and width of 4. What is its area?"
]

for question in questions:

    prompt = f"""{question} Let's think step by step."""

    response = generator(
        prompt,
        max_new_tokens=150,
        do_sample=False
    )

    print("\n" + "=" * 70)
    print(response[0]["generated_text"])
````

This demonstrates how the same reasoning prompt can be reused across different problem types.

---

## 📝 Build a Reusable CoT Prompt Template

````python
cot_template = """
Question:
{question}

Let's think step by step before giving the final answer.
"""
````

Instead of writing reasoning prompts manually, we can create a reusable template for different tasks.

---

## 🚀 Use the Prompt Template

````python
prompt = cot_template.format(
    question="A car travels 240 km in 4 hours. What is its average speed?"
)

response = generator(
    prompt,
    max_new_tokens=150,
    do_sample=False
)

print(response[0]["generated_text"])
````

Prompt templates make reasoning prompts more consistent and reusable across applications.
