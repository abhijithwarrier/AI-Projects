**PYTHON SCRIPT TO *CONNECT MULTIPLE PROMPTS TO BUILD MORE STRUCTURED AND RELIABLE AI WORKFLOWS*. 🔗🧠🤖**

A single prompt can handle many AI tasks, but complex problems often involve several different steps. Asking a Large Language Model to understand information, analyze it, generate something, and refine the result—all within one large prompt—can make the workflow harder to control.

**Prompt Chaining** solves this by breaking a complex task into smaller prompts that run sequentially. The output from one prompt becomes the input for the next, creating a structured AI workflow.

In this project, we'll build a simple Prompt Chaining pipeline using Python and an instruction-following language model.

## 📦 Import Required Libraries

````python
from transformers import pipeline
````

We'll use a Hugging Face instruction-following model to execute each step in our prompt chain.

---

## 🤖 Load an Instruction-Following Model

````python
generator = pipeline(
    "text-generation",
    model="google/gemma-2b-it"
)
````

Each prompt in the chain will be processed separately, allowing us to control what happens at every stage.

---

## 📝 Create the Input Text

Let's start with a short piece of information that will move through our prompt chain.

````python
article = """
Artificial Intelligence is transforming industries by allowing computers
to perform tasks that traditionally required human intelligence. Modern AI
systems can analyze data, recognize patterns, generate content, and assist
people with increasingly complex tasks.
"""
````

Instead of asking the model to perform several operations at once, we'll divide the workflow into individual stages.

---

## 🔗 Step 1 — Summarise the Content

The first prompt extracts the most important information.

````python
summary_prompt = f"""
Summarize the following text in two sentences.

Text:
{article}
"""
````

Generate the summary:

````python
summary_response = generator(
    summary_prompt,
    max_new_tokens=100,
    do_sample=False
)

summary = summary_response[0]["generated_text"]
````

The output from this stage becomes the input for the next prompt.

---

## 🔗 Step 2 — Extract the Main Topic

Now we'll use the generated summary rather than the original article.

````python
topic_prompt = f"""
Identify the main topic from the following summary.

Return only the topic.

Summary:
{summary}
"""
````

Generate the topic:

````python
topic_response = generator(
    topic_prompt,
    max_new_tokens=40,
    do_sample=False
)

topic = topic_response[0]["generated_text"]
````

We now have a simple chain:

```python
Article → Summary → Main Topic
```

---

## 🔗 Step 3 — Generate a Question

The next prompt uses the summary to create an educational question.

````python
question_prompt = f"""
Create one educational question based on the following summary.

Summary:
{summary}
"""
````

````python
question_response = generator(
    question_prompt,
    max_new_tokens=60,
    do_sample=False
)

question = question_response[0]["generated_text"]
````

Instead of trying to summarise and generate a question simultaneously, each stage has one clear responsibility.

---

## 🔗 Step 4 — Generate the Final Answer

The generated question can now become the input for another prompt.

````python
answer_prompt = f"""
Answer the following question using the provided summary.

Summary:
{summary}

Question:
{question}
"""
````

````python
answer_response = generator(
    answer_prompt,
    max_new_tokens=120,
    do_sample=False
)

answer = answer_response[0]["generated_text"]
````

Our complete chain now looks like:

```python
Article
   │
   ▼
Summarize
   │
   ▼
Extract Topic
   │
   ▼
Generate Question
   │
   ▼
Generate Answer
```

---

## 🧩 Create a Reusable Prompt Function

Repeated model calls can be wrapped inside a helper function.

````python
def generate_response(prompt, max_tokens=100):
    response = generator(
        prompt,
        max_new_tokens=max_tokens,
        do_sample=False
    )
    return response[0]["generated_text"]
````

Now each stage becomes easier to execute:

````python
summary = generate_response(summary_prompt)
topic = generate_response(topic_prompt, 40)
question = generate_response(question_prompt, 60)
answer = generate_response(answer_prompt, 120)
````

This makes the chain cleaner and easier to extend.

---

## 🛠️ Why Not Use One Large Prompt?

We could ask the model to summarise the article, identify the topic, generate a question, and answer it in a single prompt.

But separating the workflow provides several advantages:

- Each stage has a clear responsibility.
- Intermediate outputs can be inspected.
- Individual prompts can be improved independently.
- Errors are easier to identify.
- Outputs can be reused by multiple downstream steps.
- Complex workflows become easier to maintain.

The trade-off is that additional model calls can increase latency and computational cost, so Prompt Chaining is most useful when the task genuinely benefits from multiple stages.

```
Initial Input
      │
      ▼
   Prompt 1
      │
      ▼
Intermediate Output
      │
      ▼
   Prompt 2
      │
      ▼
Intermediate Output
      │
      ▼
   Prompt 3
      │
      ▼
 Final Output
```

The important idea is that **each output becomes context for the next stage**.

## 🧠 Key Takeaways

- Prompt Chaining breaks complex AI tasks into smaller stages.
- The output of one prompt becomes input for another.
- Each stage can focus on one specific responsibility.
- Intermediate results make workflows easier to inspect and debug.
- Prompt chains can improve control and maintainability for complex tasks.
- More stages also mean more model calls, so chains should be designed deliberately.
