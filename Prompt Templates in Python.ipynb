As AI applications grow, writing prompts manually for every request quickly becomes repetitive and error-prone. A better approach is to create **Prompt Templates**—reusable prompt structures with placeholders that can be filled dynamically.

Prompt Templates help developers maintain consistency, reduce duplication, and make AI workflows easier to manage. They are especially useful in chatbots, summarisation systems, classification pipelines, RAG applications, and automated content generation.

In this project, we'll build reusable Prompt Templates using regular Python strings.

We'll learn how to:

- Create reusable prompt templates
- Insert dynamic values into prompts
- Build templates for different AI tasks
- Reuse the same structure across multiple inputs
- Understand why Prompt Templates are important in production AI systems

## 📦 Import Required Libraries

````python
from transformers import pipeline
````

We'll use a Hugging Face instruction-following model to test the prompts generated by our templates.

---

## 🤖 Load an Instruction-Following Model

````python
generator = pipeline(
    "text-generation",
    model="google/gemma-2b-it"
)
````

The model will receive dynamically generated prompts and produce responses based on the instructions inside each template.

---

## 🧩 Create a Simple Prompt Template

````python
prompt_template = """
Explain the following topic in simple terms.

Topic: {topic}
"""
````

The `{topic}` placeholder can be replaced with different values without rewriting the entire prompt.

---

## 🔄 Insert Dynamic Values

````python
topic = "Machine Learning"

prompt = prompt_template.format(
    topic=topic
)

print(prompt)
````

The resulting prompt becomes:

```
Explain the following topic in simple terms.

Topic: Machine Learning
```

---

## 🚀 Generate a Response

````python
response = generator(
    prompt,
    max_new_tokens=100,
    do_sample=False
)

print(response[0]["generated_text"])
````

The same template can now be reused for any topic.

---

## 🔁 Reuse the Template for Multiple Topics

````python
topics = [
    "Artificial Intelligence",
    "Machine Learning",
    "Deep Learning",
    "Large Language Models"
]

for topic in topics:
    prompt = prompt_template.format(
        topic=topic
    )

    response = generator(
        prompt,
        max_new_tokens=100,
        do_sample=False
    )

    print("\n" + "=" * 60)
    print("Topic:", topic)
    print(response[0]["generated_text"])
````

Instead of writing four separate prompts, we reuse one template with different values.

---

## 📝 Create a Summarization Template

````python
summary_template = """
Summarize the following text in {sentences} sentences.

Text:
{text}
"""
````

This template contains two placeholders:

- `{sentences}` → desired summary length
- `{text}` → document to summarise

---

## 📄 Use the Summarisation Template

````python
document = (
    "Python is a widely used programming language known for its simple syntax, "
    "large ecosystem, and strong support for automation, data science, "
    "machine learning, and artificial intelligence."
)

prompt = summary_template.format(
    sentences=2,
    text=document
)

response = generator(
    prompt,
    max_new_tokens=80,
    do_sample=False
)

print(response[0]["generated_text"])
````

Prompt Templates make it easy to control both the content and structure of the request.

---

## 🏷️ Create a Classification Template

````python
classification_template = """
Classify the following customer review as Positive, Negative, or Neutral.

Review:
{review}

Sentiment:
"""
````

---

## 📊 Classify Multiple Reviews

````python
reviews = [
    "The product quality was excellent.",
    "The delivery was extremely late.",
    "The product is okay but nothing special."
]

for review in reviews:
    prompt = classification_template.format(
        review=review
    )

    response = generator(
        prompt,
        max_new_tokens=20,
        do_sample=False
    )

    print("\nReview:", review)
    print("Response:", response[0]["generated_text"])
````

The same classification instructions are consistently applied to every review.

---

## 🎭 Create a Role-Based Prompt Template

````python
role_template = """
You are a {role}.

Explain the following topic for a {audience} audience.

Topic:
{topic}
"""
````

---

## 🧠 Generate Different Prompt Variations

````python
prompt = role_template.format(
    role="Machine Learning Engineer",
    audience="beginner",
    topic="Neural Networks"
)

response = generator(
    prompt,
    max_new_tokens=120,
    do_sample=False
)

print(response[0]["generated_text"])
````

By changing the template variables, the same prompt structure can support different roles, audiences, and topics.

---

## 🛠️ Build a Reusable Prompt Function

````python
def create_prompt(template, **kwargs):
    return template.format(**kwargs)
````

Now prompts can be generated with a reusable helper function.

````python
prompt = create_prompt(
    prompt_template,
    topic="Retrieval-Augmented Generation"
)

print(prompt)
````

This approach becomes especially useful when AI applications contain many different prompt templates.
