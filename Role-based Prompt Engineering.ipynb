**PYTHON SCRIPT *TO GUIDE LARGE LANGUAGE MODELS BY ASSIGNING THEM A CLEAR ROLE, EXPERTISE, AND AUDIENCE*. 🎭🧠💬**

Large Language Models can respond to the same question in very different ways depending on how the prompt is written. One simple but powerful technique is **Role-based Prompt Engineering**, where we explicitly tell the model who it should act as before giving it the actual task.

Instead of asking:

> Explain neural networks.
>

we can ask:

> You are a Machine Learning instructor. Explain neural networks to a beginner using simple language and an everyday analogy.
>

The second prompt gives the model much more context about the expected expertise, tone, complexity, and style of the response.

In this project, we'll explore how assigning different roles can influence LLM responses.

## 📦 Import Required Libraries

````python
from transformers import pipeline
````

We'll use a Hugging Face instruction-following language model to compare responses generated from different role-based prompts.

---

## 🤖 Load an Instruction-Following Model

````python
generator = pipeline(
    "text-generation",
    model="google/gemma-2b-it"
)
````

Instruction-tuned models respond well when prompts clearly specify the role, task, and expected style.

---

## 📝 Start with a Regular Prompt

````python
regular_prompt = """
Explain Artificial Intelligence.
"""
````

This gives the model very little guidance beyond the actual topic.

---

## 🚀 Generate the Regular Response

````python
response = generator(
    regular_prompt,
    max_new_tokens=120,
    do_sample=False
)

print(response[0]["generated_text"])
````

The model decides the tone, level of detail, and explanation style on its own.

---

## 🎭 Add a Role to the Prompt

````python
role_prompt = """
You are an Artificial Intelligence teacher.

Explain Artificial Intelligence to a beginner using simple language and one real-world example.
"""
````

Now the model knows:

- who it should behave as,
- who the audience is,
- how complex the explanation should be,
- and what type of response is expected.

---

## 🚀 Generate the Role-Based Response

````python
response = generator(
    role_prompt,
    max_new_tokens=150,
    do_sample=False
)

print(response[0]["generated_text"])
````

The response should now be more aligned with the assigned teaching role and beginner audience.

---

## 🔄 Compare Different Roles

The same topic can be explained differently depending on the role assigned to the model.

````python
roles = [
    "Machine Learning Teacher",
    "Senior AI Engineer",
    "Technical Writer",
    "Business Consultant"
]

topic = "Large Language Models"

for role in roles:
    prompt = f"""
You are a {role}.

Explain {topic} clearly in a way appropriate for your role.
"""

    response = generator(
        prompt,
        max_new_tokens=140,
        do_sample=False
    )

    print("\n" + "=" * 70)
    print("Role:", role)
    print(response[0]["generated_text"])
````

A technical writer may focus on clarity, while an AI engineer may provide a more technical explanation. A business consultant may instead focus on business impact and practical applications.

## 👥 Adapt the Same Role for Different Audiences

Role prompting becomes even more powerful when the target audience is also defined.

````python
template = """
You are a {role}.

Explain {topic} to a {audience}.

Keep the explanation appropriate for their level of understanding.
"""
````

We can dynamically change the audience:

````python
prompt = template.format(
    role="Machine Learning Instructor",
    topic="Neural Networks",
    audience="complete beginner"
)
````

The model can now adapt the same topic for beginners, students, developers, executives, or domain experts.

---

## 🧩 Build a Reusable Role Prompt Function

````python
def create_role_prompt(role, task, audience, style):
    return f"""
You are a {role}.

Task:
{task}

Audience:
{audience}

Response Style:
{style}
"""
````

Now we can generate consistent role-based prompts dynamically.

````python
prompt = create_role_prompt(
    role="Python Instructor",
    task="Explain decorators in Python.",
    audience="intermediate Python developers",
    style="clear, concise, and practical"
)
````

This makes role prompting much easier to integrate into larger AI applications.

```
support_prompt = """
```

Changing the role changes how the model approaches the same underlying task.

## 📈 How Role-based Prompting Works

```
User Task
     │
     ▼

Assign Role
     │
     ▼

Define Audience
     │
     ▼

Specify Style / Constraints
     │
     ▼

Large Language Model
     │
     ▼

Context-Aware Response
```

A good role prompt provides the model with enough context to understand not only **what** to do, but also **how** it should respond.

## Conclusion

Role-based Prompt Engineering gives developers greater control over how Large Language Models respond. By defining the model's role, audience, task, and response style, we can transform a generic LLM into a more focused assistant suited to a specific use case.

The role doesn't magically give the model new knowledge, but it helps guide how existing knowledge is selected, organised, and communicated. This makes Role-based Prompt Engineering particularly useful for AI tutors, coding assistants, support systems, enterprise copilots, and specialised AI agents.
