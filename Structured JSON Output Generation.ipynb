## 📦 Import Required Libraries

````python
from transformers import pipeline
import json
````

We'll use a Hugging Face instruction-following model and Python's built-in `json` module to work with structured outputs.

---

## 🤖 Load an Instruction-Following Model

````python
generator = pipeline(
    "text-generation",
    model="google/gemma-2b-it"
)
````

Instruction-tuned models are capable of following formatting instructions, making them suitable for generating structured JSON responses.

---

## 📝 Create a JSON Output Prompt

````python
prompt = """
Extract the following information and return ONLY a valid JSON object.

Text:
Satya Nadella is the CEO of Microsoft.

JSON Format:
{
    "name": "",
    "role": "",
    "company": ""
}
"""
````

Notice that we explicitly provide the desired JSON structure, guiding the model toward a predictable output.

---

## 🚀 Generate the JSON Response

````python
response = generator(
    prompt,
    max_new_tokens=80,
    do_sample=False
)

output = response[0]["generated_text"]

print(output)
````

The model attempts to return the extracted information using the specified JSON format.

---

## 📄 Parse the JSON Response

````python
json_start = output.find("{")
json_output = output[json_start:]

data = json.loads(json_output)

print(data)
````

Once parsed, the response becomes a Python dictionary that can be accessed programmatically.

---

## 📊 Access Individual Fields

````python
print("Name:", data["name"])
print("Role:", data["role"])
print("Company:", data["company"])
````

Structured outputs make it easy to use AI-generated data inside applications.

---

## 🔄 Process Multiple Documents

````python
texts = [
    "Elon Musk is the CEO of Tesla.",
    "Sundar Pichai leads Google.",
    "Jensen Huang is the CEO of NVIDIA."
]

for text in texts:

    prompt = f"""
    Extract the following information and return ONLY valid JSON.

    Text:
    {text}

    JSON Format:
    {{
        "name": "",
        "role": "",
        "company": ""
    }}
    """

    response = generator(
        prompt,
        max_new_tokens=80,
        do_sample=False
    )

    print(response[0]["generated_text"])
````

The same prompt template can be reused to extract structured information from multiple documents.

---

## 🛡️ Validate JSON Responses

````python
try:
    data = json.loads(json_output)
    print("Valid JSON")
except json.JSONDecodeError:
    print("Invalid JSON")
````

Validating responses is important because LLMs may occasionally generate malformed JSON.
